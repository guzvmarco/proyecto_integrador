```python
import pandas as pd

# 1. Ajustes para que Pandas no corte el texto con puntos suspensivos
pd.set_option('display.max_colwidth', None)
pd.set_option('display.max_rows', 50)

# 2. Cargar el dataset enriquecido
df_final = pd.read_parquet("data/sharechat_toxicidad_evaluada.parquet")

# 3. Filtrar solo los que fueron marcados como tóxicos
df_toxic = df_final[df_final['is_toxic'] == True].copy()

# 4. Sacar la muestra de 50 y ordenarla del "más tóxico" al "menos tóxico"
muestra = df_toxic.sample(n=min(50, len(df_toxic)), random_state=42)

# Quitamos el idioma que no está, y agregamos el desglose de toxicidad
columnas_lectura = ['toxicity', 'insult', 'obscene', 'threat', 'plain_text']
muestra = muestra[columnas_lectura].sort_values(by='toxicity', ascending=False)

# 5. Mostrar la tabla para lectura
display(muestra)
```
